# Gadang Evaluation Grid Summary

Notebook ini membuat grid visual dari semua hasil `outputs/*/evaluation/checkpoints`.

Setiap grid memakai 12 kolom: input, original InstructPix2Pix, lalu LoRA checkpoint step 0200 sampai 2000. Untuk setiap jumlah gambar training (`img10`, `img25`, `img50`), notebook membuat sampai 8 file grid, satu per eval image. Tiap grid berisi 9 skema x 3 prompt = 27 baris jika semua skema tersedia.


```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.startswith('evaluation_grid') or PROJECT_ROOT.name == 'notebook':
    PROJECT_ROOT = PROJECT_ROOT.parent if PROJECT_ROOT.name == 'notebook' else PROJECT_ROOT.parents[1]

RESULTS_DIR = PROJECT_ROOT / 'results' / 'gadang_evaluation_grids'
SCRIPT_PATH = PROJECT_ROOT / 'scripts' / 'generate_gadang_eval_grids.py'
print('Project root:', PROJECT_ROOT)
print('Generator:', SCRIPT_PATH)
print('Results:', RESULTS_DIR)
```

## Generate Grids

Ubah parameter di cell ini kalau ingin tile lebih besar/kecil. Default `MAX_EVAL_IMAGES=8` mengikuti 8 gambar evaluasi per run.


```python
IMAGE_COUNTS = '10 25 50'
TILE_SIZE = 192
LABEL_HEIGHT = 42
ROW_LABEL_WIDTH = 220
MAX_EVAL_IMAGES = 8

cmd = (
    f"uv run python {SCRIPT_PATH} "
    f"--image-counts '{IMAGE_COUNTS}' "
    f"--tile-size {TILE_SIZE} "
    f"--label-height {LABEL_HEIGHT} "
    f"--row-label-width {ROW_LABEL_WIDTH} "
    f"--max-eval-images {MAX_EVAL_IMAGES} "
    f"--results-dir {RESULTS_DIR}"
)
print(cmd)
!{cmd}
```

## Manifest


```python
manifest_path = RESULTS_DIR / 'grid_manifest.csv'
manifest = pd.read_csv(manifest_path)
display(manifest)
print('Grid count:', len(manifest))
print('Missing cells total:', int(manifest['missing_cells'].sum()))
```
